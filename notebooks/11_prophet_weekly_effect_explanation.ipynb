# Prophet Weekly Effect Explanation

Prophet models weekly seasonality as an adjustment relative to the baseline trend.

For Saturday:

`weekly_effect(Saturday) ~= Saturday revenue - baseline revenue`

`~= GBP 0 - GBP 27,009`

`~= -GBP 27,009`

The weekly effect is not the actual Saturday revenue. It is a seasonal adjustment.
With additive seasonality, Prophet calculates:

`Forecast = trend + weekly + yearly + holidays`

Because the historical data contains almost no Saturday revenue, the learned Saturday adjustment is strongly negative.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

daily = pd.read_csv(
    PROJECT_ROOT / 'data' / 'processed' / 'daily_revenue.csv',
    parse_dates=['date'],
).sort_values('date')

daily['weekday'] = daily['date'].dt.day_name()
baseline_revenue = daily['net_revenue'].mean()
saturday_revenue = daily.loc[
    daily['weekday'].eq('Saturday'),
    'net_revenue',
].mean()

weekly_effect_saturday = saturday_revenue - baseline_revenue

print(f'Baseline revenue: GBP {baseline_revenue:,.2f}')
print(f'Saturday revenue: GBP {saturday_revenue:,.2f}')
print(f'Weekly effect (Saturday): GBP {weekly_effect_saturday:,.2f}')
```

```python
summary = pd.DataFrame({
    'Metric': [
        'Overall baseline revenue',
        'Average Saturday revenue',
        'Saturday weekly effect',
    ],
    'Value': [
        baseline_revenue,
        saturday_revenue,
        weekly_effect_saturday,
    ],
})
summary
```

## Interpretation

The negative value does not mean the business has negative Saturday sales. It means Saturday is estimated to be below the overall baseline.

The final forecast can still become negative because additive components are allowed to sum below zero.
